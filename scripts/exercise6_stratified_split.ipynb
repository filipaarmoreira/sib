# Exercise 6.2 – stratified_train_test_split on iris

```python
import numpy as np
from si.io.csv_file import read_csv
from si.model_selection.split import train_test_split, stratified_train_test_split

iris = read_csv('../datasets/iris/iris.csv', sep=',', features=True, label=True)
train, test = stratified_train_test_split(iris, test_size=0.2, random_state=42)
print(train.shape(), test.shape())
```

```python
print('train:', dict(zip(*np.unique(train.y, return_counts=True))))
print('test :', dict(zip(*np.unique(test.y, return_counts=True))))
```

Comparison with the (non-stratified) train_test_split:

```python
train, test = train_test_split(iris, test_size=0.2, random_state=42)
print('test :', dict(zip(*np.unique(test.y, return_counts=True))))
```

```python
from si.models.knn_classifier import KNNClassifier

train, test = stratified_train_test_split(iris, test_size=0.2, random_state=42)
knn = KNNClassifier(k=3).fit(train)
print('Accuracy:', knn.score(test))
```
