## Exercício 1 – primeira camada à mão
Rede 2 → 3 → 1. Entrada x = [1, 2], pesos W (2×3) e bias b.

```python
import numpy as np

x = np.array([[1, 2]])                 # (1, 2)
W = np.array([[0.1, 0.2, -0.1],
              [0.3, -0.2, 0.4]])       # (2, 3)
b = np.array([[0.1, 0, -0.1]])         # (1, 3)

z = np.dot(x, W) + b
print(z)  # esperado: [[ 0.8 -0.2  0.6]]
```

## Exercício 2 – rede 2 → 3 → 1 com dados aleatórios
5 amostras, 2 features, pesos em [-0.5, 0.5[ e bias a zero.

```python
np.random.seed(42)

X = np.random.rand(5, 2)
print("X:", X.shape)

# camada 1: 2 -> 3
W1 = np.random.rand(2, 3) - 0.5
b1 = np.zeros((1, 3))
H1 = np.dot(X, W1) + b1
print("W1:", W1.shape, "b1:", b1.shape, "H1:", H1.shape)

# camada 2: 3 -> 1
W2 = np.random.rand(3, 1) - 0.5
b2 = np.zeros((1, 1))
out = np.dot(H1, W2) + b2
print("W2:", W2.shape, "b2:", b2.shape, "out:", out.shape)
out
```

## Exercício 3 – o mesmo com a `DenseLayer` do pacote

```python
from si.neural_networks.layers import DenseLayer

layer1 = DenseLayer(3, (2,)).initialize()
layer2 = DenseLayer(1, layer1.output_shape()).initialize()

H1 = layer1.forward_propagation(X)
out = layer2.forward_propagation(H1)
print("H1:", H1.shape, "out:", out.shape)
print("parâmetros:", layer1.parameters() + layer2.parameters())  # (2*3+3) + (3*1+1) = 13
```
