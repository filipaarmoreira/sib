```python
import os
import matplotlib.pyplot as plt
from si.io.csv_file import read_csv
from si.io.csv_file import read_csv
from si.model_selection.split import train_test_split
from si.models.logistic_regression import LogisticRegression
```

```python
ds = read_csv('../datasets/breast_bin/breast-bin.csv', sep=',', features=False, label=True)
train, test = train_test_split(ds, test_size=0.3, random_state=42)
```

```python
model = LogisticRegression(l2_penalty=1, alpha=0.001, max_iter=2000)
model.fit(train)
print("Accuracy:", model.score(test))
print("Cost:", model.cost(test))
```
