**Exercise 2** – dropna, fillna, remove_by_index

```python
from copy import deepcopy

iris_na = read_csv('../datasets/iris/iris_missing_data.csv', sep=',', features=True, label=True)
print('NaNs:', np.isnan(iris_na.X.astype(float)).sum(), 'shape:', iris_na.shape())
```

```python
ds = deepcopy(iris_na).dropna()
print(ds.shape(), len(ds.y))
```

```python
ds = deepcopy(iris_na).fillna(0.0)
print(ds.shape(), np.isnan(ds.X).sum())
```

```python
ds = deepcopy(iris_na).fillna('mean')
print(np.isnan(ds.X).sum())
```

```python
ds = deepcopy(iris_na).fillna('median')
print(np.isnan(ds.X).sum())
```

```python
ds = deepcopy(iris).remove_by_index(0)
print(ds.shape(), len(ds.y))
```
