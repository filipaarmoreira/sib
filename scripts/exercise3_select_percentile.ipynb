# Exercise 3.3 – SelectPercentile on iris

```python
from si.io.csv_file import read_csv
from si.feature_selection.select_percentile import SelectPercentile

iris = read_csv('../datasets/iris/iris.csv', sep=',', features=True, label=True)
selector = SelectPercentile(percentile=50).fit(iris)
print('F:', selector.F)
print('p:', selector.p)
```

```python
new = selector.transform(iris)
print(new.features, new.shape())
```
