# Exercise 7.3 – KNNRegressor on cpu.csv

```python
from si.io.csv_file import read_csv
from si.model_selection.split import train_test_split
from si.models.knn_regressor import KNNRegressor

cpu = read_csv('../datasets/cpu/cpu.csv', sep=',', features=True, label=True)
train, test = train_test_split(cpu, test_size=0.2, random_state=42)
print(train.shape(), test.shape())
```

```python
knn = KNNRegressor(k=3).fit(train)
predictions = knn.predict(test)
print(predictions[:10])
print(test.y[:10])
```

```python
print('RMSE:', knn.score(test))
```

```python
for k in [1, 3, 5, 7, 9]:
    print(k, KNNRegressor(k=k).fit(train).score(test))
```
